```python
#3
import csv

with open("ipl_matches.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Writing column headings
    writer.writerow(["Match", "Team1", "Team2", "Winner"])

    # Writing match details
    writer.writerow([1, "CSK", "MI", "CSK"])
    writer.writerow([2, "RCB", "KKR", "KKR"])
    writer.writerow([3, "MI", "SRH", "MI"])
    writer.writerow([4, "CSK", "RCB", "RCB"])
    writer.writerow([5, "KKR", "SRH", "KKR"])

print("IPL CSV file created successfully!")
```

```python
import csv

with open("ipl_matches.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print("Match", row["Match"], "- Winner:", row["Winner"])
```
