```python
#2
playlist_prices = {
    "Top Hits": 199,
    "Bollywood Hits": 149,
    "English Songs": 179,
    "Workout Mix": 99,
    "Romantic Songs": 129
}

def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price

update_playlist_price("Top Hits", 249)

print("Updated Playlist Prices:")
print(playlist_prices)
```

```python
#4
set1 = {"Dominos", "McDonald's", "Burger King", "Pizza Hut"}

set2 = {"McDonald's", "Pizza Hut", "Subway", "KFC"}

union_set = set1.union(set2)

intersection_set = set1.intersection(set2)

print("Zomato Restaurants:", set1)
print("Swiggy Restaurants:", set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```
