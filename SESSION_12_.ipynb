```python
#1
add_gst = lambda price: price + (price * 0.18)

print("Price after GST:", add_gst(100))
print("Price after GST:", add_gst(250))
print("Price after GST:", add_gst(500))
```

```python
#2
song_titles = [
    ' shape OF you ',
    '  believer',
    'KESARIYA ',
    ' blinding LIGHTS '
]

cleaned_songs = list(map(lambda song: song.strip().title(), song_titles))

print(cleaned_songs)
```

```python
#3
products = [
    'Samsung Mobile',
    'Shoes',
    'Laptop',
    'Smart Watch',
    'Headphones',
    'Smart TV',
    'Sunglasses'
]

filtered_products = list(filter(lambda product: product.lower().startswith('s'), products))

print(filtered_products)
```

```python
#4
from functools import reduce

order_amounts = [120, 340, 560, 80]

total_bill = reduce(lambda total, amount: total + amount, order_amounts)

print("Total Zomato Bill:", total_bill)
```

```python
#5
from functools import reduce

numbers = [40, 60, 80, 120]

doubled_numbers = list(map(lambda x: x * 2, numbers))

filtered_numbers = list(filter(lambda x: x > 100, doubled_numbers))

total = reduce(lambda x, y: x + y, filtered_numbers)

print("Original Numbers:", numbers)
print("Doubled Numbers:", doubled_numbers)
print("Filtered Numbers:", filtered_numbers)
print("Final Total:", total)
```
