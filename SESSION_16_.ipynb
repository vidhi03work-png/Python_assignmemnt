```python
#1
movies = [
    "Jawan",
    "Pathaan",
    "Animal",
    "Stree 2",
    "Bhool Bhulaiyaa 3"
]

movie_iterator = iter(movies)

print(next(movie_iterator))
print(next(movie_iterator))
print(next(movie_iterator))
print(next(movie_iterator))
print(next(movie_iterator))
```

```python
#2
playlist = [
    "Kesariya",
    "Perfect",
    "Heeriye",
    "Tum Hi Ho",
    "Apna Bana Le",
    "Chaleya"
]

for position, song in enumerate(playlist, start=1):
    print(position, ".", song)
```

```python
#3
food_items = ["Pizza", "Burger", "Pasta", "Sandwich", "Dosa"]

prices = [250, 120, 180, 100, 80]

for food, price in zip(food_items, prices):
    print(food, "-", "₹" + str(price))
```

```python
#4
def insta_posts_generator(posts):

    for post in posts:
        yield post

posts = [
    "Enjoying my weekend!",
    "New beginnings",
    "Travel diaries",
    "Keep smiling!",
    "Dream big"
]

post_generator = insta_posts_generator(posts)

try:
    print(next(post_generator))
    print(next(post_generator))
    print(next(post_generator))
    print(next(post_generator))
    print(next(post_generator))

except StopIteration:
    print("All posts have been displayed.")
```

```python
#5
def cashback_generator(transactions):

    for amount in transactions:
        cashback = amount * 0.05
        yield cashback

transactions = [500, 1000, 200, 1500, 800]

cashback = cashback_generator(transactions)

for value in cashback:
    print("Cashback: ₹", value)
```
