```python
#1
order_amounts = [120, 250, 90, 310, 150]

total = 0

for amount in order_amounts:
    total = total + amount

print("Total Order Value:", total)
```

```python
#2
cricket_scores = [45, 78, 102, 34, 67, 89]

i = 0

while i < len(cricket_scores):

    if cricket_scores[i] > 100:
        break

    print(cricket_scores[i])

    i = i + 1
```

```python
#3
item_prices = [299, 499, 199, 999, 149]

total = 0

for price in item_prices:

    if price < 200:
        continue

    total = total + price

print("Total Cart Value:", total)
```

```python
#4
favorite_songs = [
    'Kesariya',
    'Believer',
    'Shape of You',
    'Blinding Lights',
    'Excuses'
]

for position, song in enumerate(favorite_songs, start=1):
    print(position, song)
```

```python
#5
follower_counts = [120, 1500, 23000, 800, 45000]

for followers in follower_counts:

    if followers < 1000:
        print(followers, "Micro")

    elif followers <= 10000:
        print(followers, "Influencer")

    else:
        print(followers, "Celebrity")
```
