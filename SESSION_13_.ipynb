```python
#1
def reverse_string(s):

    if len(s) == 0:
        return s

    return reverse_string(s[1:]) + s[0]


result = reverse_string("hello")

print("Original String:", "hello")
print("Reversed String:", result)
```

```python
#2
def sum_playlist_durations(durations):


    if len(durations) == 0:
        return 0


    return durations[0] + sum_playlist_durations(durations[1:])


song_durations = [180, 240, 200, 150, 210]

total_duration = sum_playlist_durations(song_durations)

print("Total Playlist Duration:", total_duration, "seconds")
```

```python
#3
count = 10

def update_count():
    count = 5
    print("Inside:", count)

update_count()

print("Outside:", count)
```

```python
#1 The variable count = 10 is a global variable because it is declared outside the function.

#2 The variable count = 5 is a local variable because it is declared inside the function.

#3 When the function is called, it prints Inside: 5 because it uses the local variable.

#4 Outside the function, it prints Outside: 10 because the global variable remains unchanged.

#5 The local variable does not change the global variable.
```

```python
#4
def count_likes(posts):

    total = 0

    if isinstance(posts, dict):

        total += posts.get("likes", 0)

        for key, value in posts.items():

            if key != "likes":
                total += count_likes(value)

    elif isinstance(posts, list):

        for item in posts:
            total += count_likes(item)

    return total


instagram_posts = {
    "post1": {
        "likes": 100,
        "replies": [
            {"likes": 20},
            {"likes": 15}
        ]
    },

    "post2": {
        "likes": 200,
        "replies": [
            {"likes": 30},
            {"likes": 25}
        ]
    }
}


total = count_likes(instagram_posts)

print("Total Instagram Likes:", total)
```
